```python
import os
import random
import pandas as pd

DATASET_PATH = "Cows datasets"

DISEASES = {

    "healthy": {
        "symptoms": [
            "healthy",
            "active",
            "normal appetite",
            "normal temperature",
            "bright eyes",
            "good body condition",
            "normal respiration",
            "normal heart rate",
            "no visible lesions",
            "normal behavior"
        ],
        "severity": ["Low"]
    },

    "lumpy": {
        "symptoms": [
            "fever",
            "skin nodules",
            "swollen lymph nodes",
            "loss of appetite",
            "nasal discharge",
            "eye discharge",
            "weight loss",
            "reduced milk production",
            "lameness",
            "depression"
        ],
        "severity": ["Medium", "High", "Critical"]
    },

    "foot-and-mouth": {
        "symptoms": [
            "mouth ulcers",
            "hoof lesions",
            "excessive salivation",
            "drooling",
            "fever",
            "lameness",
            "reduced appetite",
            "blisters",
            "difficulty walking",
            "reduced milk production"
        ],
        "severity": ["Medium", "High", "Critical"]
    }

}

rows = []

for disease in os.listdir(DATASET_PATH):

    folder = os.path.join(DATASET_PATH, disease)

    if not os.path.isdir(folder):
        continue

    for image in os.listdir(folder):

        if image.lower().endswith((".jpg",".jpeg",".png")):

            n = random.randint(2,5)

            symptoms = random.sample(
                DISEASES[disease]["symptoms"],
                n
            )

            severity = random.choice(
                DISEASES[disease]["severity"]
            )

            rows.append({
                "image": image,
                "disease": disease,
                "symptoms": ", ".join(symptoms),
                "severity": severity
            })

df = pd.DataFrame(rows)

df.to_csv("symptoms.csv", index=False)

print(df.head())

print(f"\nDataset créé : {len(df)} images")
```
